```python
df = df.dropna(subset=["id"])
```

```python
df = df[~df["invoice"].astype(str).str.startswith("c")]
```

```python
df = df[   (df["quan"] > 0  &  df["unit"] > 0)   ]
```

```python
print(df.shape)
print(df["id"].unique())
```

```python
c_date = pd.Timestamp("2011-10-31")
```

```python
past = df[ df["inv"] <= c_date].copy()

future = df[
    (df["inv"] > c_date) &
    (df["inv"] <= c_date + pd.Timedelta(days=30))
].copy()
```

```python
print("Past:", past["InvoiceDate"].min(), "→", past["InvoiceDate"].max())
print("Future:", future["InvoiceDate"].min(), "→", future["InvoiceDate"].max())
```
