```python
import pandas as pd
import numpy as np

df = pd.read_csv("customer_churn.csv")

print(df.head())
print(df.shape)
print(df.info())
```

Automated EDA
--

```python
# Basic information
print("Shape:", df.shape)

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Duplicate records
print("\nDuplicates:", df.duplicated().sum())

# Statistical summary
print("\nStatistics:")
print(df.describe(include="all"))

# Data types
print("\nData Types:")
print(df.dtypes)
```

```python
# Churn distribution
print(df["Churn"].value_counts())

# Churn percentage
print(df["Churn"].value_counts(normalize=True) * 100)

# Churn by contract
print(
    df.groupby("Contract")["Churn"]
      .value_counts(normalize=True)
      .mul(100)
)
```

Generate EDA report automatically
--

```python
from ydata_profiling import ProfileReport

profile = ProfileReport(
    df,
    title="Customer Churn EDA Report",
    explorative=True
)

profile.to_file("eda_report.html")
```

Connect AI to EDA
--

```python
import json
from openai import OpenAI

client = OpenAI()

summary = {
    "shape": df.shape,
    "columns": df.columns.tolist(),
    "missing_values": df.isnull().sum().to_dict(),
    "churn_distribution": df["Churn"].value_counts().to_dict(),
    "statistics": df.describe(include="all").to_dict()
}

prompt = f"""
Act as a Senior Data Analyst.

Analyze this EDA summary:

{json.dumps(summary, default=str)}

Provide:
1. Important patterns
2. Data quality issues
3. Business insights
4. Five recommendations

Never invent information.
"""

response = client.responses.create(
    model="gpt-5.4-mini",
    input=prompt
)

print(response.output_text)
```
